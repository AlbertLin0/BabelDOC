```python
import httpx

async def translate_pdf(pdf_path: str, server_url: str = "http://localhost:8086"):
    async with httpx.AsyncClient(timeout=3600) as client:
        with open(pdf_path, "rb") as f:
            files = {"file": (pdf_path, f, "application/pdf")}
            resp = await client.post(f"{server_url}/translate", files=files)

        if resp.status_code != 200:
            print(f"翻译失败: {resp.text}")
            return

        # 读取响应头获取文件名信息
        request_id = resp.headers.get("X-Request-ID")
        mono_name = resp.headers.get("X-Mono-Filename")
        dual_name = resp.headers.get("X-Dual-Filename")
        print(f"Request-ID: {request_id}")
        print(f"Mono: {mono_name}, Dual: {dual_name}")

        # 保存 ZIP 文件
        zip_path = f"{pdf_path}_translated.zip"
        with open(zip_path, "wb") as f:
            f.write(resp.content)
        print(f"已保存: {zip_path}")

import os
print(os.getcwd())
# 运行
await translate_pdf("./tes1t.pdf")
```
